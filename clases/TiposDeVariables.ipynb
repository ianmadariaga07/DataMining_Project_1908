```python
#Rango intercuartilico
Q1 = df['Ingresos'].quantile(0.25)
Q2 = df['Ingresos'].quantile(0.50)
Q3 = df['Ingresos'].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q2 (Mediana):", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
#
```

```python
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print("Limite Inferior:", limite_inferior)
print("Limite Superior:", limite_superior)
#
```

```python
#Mostrar los registros con datos atipicos en Ingresos
df[['Edad', 'Ingresos', 'Genero']][
    (df['Ingresos'] < limite_inferior) | (df['Ingresos'] > limite_superior)
  ]
```

```python
plt.figure(figsize=(10, 2))
plt.hlines(1, Q1, Q3, linewidth=12)
plt.scatter([Q1, Q3], [1, 1], s=80)

plt.axvline(limite_inferior, color='blue', linestyle='--', linewidth=2)
plt.axvline(limite_superior, color='blue', linestyle='--', linewidth=2)

plt.text(Q1, 1.15, 'Q1', ha='center')
plt.text(Q3, 1.15, 'Q3', ha='center')
plt.text((Q1 + Q3)/2, 1.15, 'IQR', ha='center')

plt.title('Rango Intercuartílico')
plt.xlabel('Ingresos')

plt.show()
```

```python
#Asimetria
# ≈ 0  distribucion simetrica
# > 0 asimetria positiva, cargada a la derecha
# < 0 asimetria negativa, cargada a la izquierda
df[variables_numericas].skew()
```

```python
for v in variables_numericas:
  sns.histplot(df[v], kde=True)
  plt.axvline(df[v].mean(), color='red', linestyle='--', label='Media')
  plt.axvline(df[v].median(), color='green', linestyle='--', label='Mediana')
  plt.axvline(df[v].mode()[0], color='blue', linestyle='--', label='Moda')

  plt.title(f'{v} - Asimetría: {df[v].skew():.2f}')
  plt.xlabel(v)
  plt.ylabel('Frecuencia')
  plt.legend()
  plt.show()
```
